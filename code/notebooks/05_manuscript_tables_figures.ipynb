# BNC2 manuscript tables and figures

This notebook reruns the current Gibbs/GFW-based R1-R4 manuscript artefact builder. It writes the manuscript tables and figures to their dedicated output folders.

**Scope.** R1 reports the forest service and its concentration. R2 reports the effect of blue carbon. R3 reports the erosion of the forest service. R4 compares illustrative compensation costs and allocation rules.

**Important.** The R4 prices remain illustrative until an IPCC AR6 scenario, price year, and conversion to 2015 US dollars are selected.

## Reproducibility

The script uses fixed Gibbs/GFW mean rates for physical flows and propagates GSCC draws for monetary values. Spatial rate dispersion is handled only as a deterministic sensitivity. Missing FRA deforestation reports remain missing, not zero.

```python
from pathlib import Path
import runpy

root = Path.cwd().resolve()
while not (root / 'data').exists() and root != root.parent:
    root = root.parent
if not (root / 'data').exists():
    raise FileNotFoundError('Run this notebook from the repository or one of its subdirectories.')

runpy.run_path(root / 'scripts' / 'build_manuscript_artifacts.py', run_name='__main__')
```

## Outputs

The four figures are written to `results/figures/manuscript/`. The four main-text tables and the country-level erosion table are written to `results/tables/manuscript/`.

```python
import pandas as pd

tables_dir = root / 'results' / 'tables' / 'manuscript'
figures_dir = root / 'results' / 'figures' / 'manuscript'
display(pd.read_csv(tables_dir / 'Table1_global_carbon_wealth.csv'))
display(pd.read_csv(tables_dir / 'Table2_group_profiles_and_balances.csv'))
display(pd.read_csv(tables_dir / 'Table3_erosion_by_group.csv'))
display(pd.read_csv(tables_dir / 'Table4_illustrative_compensation_scenarios.csv'))
sorted(path.name for path in figures_dir.glob('*.png'))
```
